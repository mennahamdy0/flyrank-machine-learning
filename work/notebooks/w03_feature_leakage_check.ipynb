# Build the Feature Vector

The feature vector contains only variables that are available before making a recommendation.

I selected features that describe search visibility and user engagement without using any future information or label-derived fields.

```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    split="train[:5000]"
)

df = dataset.to_pandas()

print(df.shape)
df.head()
```

```python
import numpy as np

feature_df = df.copy()
```

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np

feature_df = df.copy()

# CTR
feature_df["ctr"] = np.where(
    feature_df["gsc_impressions"] > 0,
    feature_df["gsc_clicks"] / feature_df["gsc_impressions"],
    0
)

# Engagement Rate
feature_df["engagement_rate"] = np.where(
    feature_df["ga4_sessions"] > 0,
    feature_df["ga4_engaged_sessions"] / feature_df["ga4_sessions"],
    0
)

# Organic Traffic Ratio
feature_df["organic_ratio"] = np.where(
    feature_df["ga4_sessions"] > 0,
    feature_df["sessions_organic"] / feature_df["ga4_sessions"],
    0
)

# Final Feature Vector
X = feature_df[[
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position",
    "ctr",
    "engagement_rate"
]]

X = X.fillna(0)

print("Feature Vector Shape:", X.shape)
display(X.head())
```

## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

# Leakage Hunt

I intentionally tested one feature that should not be used.

Using future or label-derived information can artificially improve model performance.

Only features known before the prediction moment should remain in the final model.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

feature_df["fake_label"] = (
    feature_df["gsc_clicks"] > feature_df["gsc_clicks"].median()
).astype(int)

X_leak = X.copy()
X_leak["fake_label"] = feature_df["fake_label"]

y = feature_df["fake_label"]

tree = DecisionTreeClassifier(random_state=42)

tree.fit(X_leak, y)

pred = tree.predict(X_leak)

print("Accuracy with leakage:",
      accuracy_score(y, pred))
```

```python
X_leak = X.copy()

print("Leakage feature removed.")
```
